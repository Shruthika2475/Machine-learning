# Create a TransactionType column to classify records as cancellation, return, positive sale or other/invalid, using documented rules.

```python

def classify_transaction(row):
    invoice = row["InvoiceNo"]
    quantity = row["Quantity"]

    # Rule 1: Missing or invalid values
    if pd.isna(invoice) or pd.isna(quantity):
        return "Other/Invalid"

    invoice = str(invoice).strip()

    # Rule 2: Cancellation invoice
    if invoice.upper().startswith("C"):
        return "Cancellation"
    # Rule 3: Negative quantity
    if quantity < 0:
        return "Return"
    # Rule 4: Positive quantity
    if quantity > 0:
        return "Positive Sale"
    # Rule 5: Zero quantity
    return "Other/Invalid"

df["TransactionType"] = df.apply(classify_transaction, axis=1)
print(df[["InvoiceNo", "Quantity", "TransactionType"]].head(20))
df
```

# 	Remove exact duplicate rows from a working copy and verify how many records remain.

```python

df_clean = df.copy()
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates()
rows_after = len(df_clean)
print("Records before removing duplicates:", rows_before)
print("Records after removing duplicates:", rows_after)
print("Duplicate rows removed:", rows_before - rows_after)
```

# Investigate repeated InvoiceNo and StockCode combinations without automatically deleting them, since multiple lines may be legitimate.

```python

invoice_stock_counts = (
    df_clean.groupby(["InvoiceNo", "StockCode"], dropna=False)
    .size()
    .reset_index(name="OccurrenceCount")
)
repeated_combinations = invoice_stock_counts[
    invoice_stock_counts["OccurrenceCount"] > 1
].sort_values("OccurrenceCount", ascending=False)
print("Number of repeated InvoiceNo-StockCode combinations:",
      len(repeated_combinations))
print(repeated_combinations)
```

```python
repeated_records = df_clean.merge(
    repeated_combinations[["InvoiceNo", "StockCode"]],
    on=["InvoiceNo", "StockCode"],
    how="inner"
)
print(repeated_records.sort_values(["InvoiceNo", "StockCode"]))
```
